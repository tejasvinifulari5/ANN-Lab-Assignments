```python
# Program to implement AND and OR logic gates using McCulloch-Pitts Neural Network

# Input combinations for two binary variables
inputs = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

# Function to implement McCulloch-Pitts neuron
def mcp_neuron(inputs, weights, threshold):
    # Calculate the weighted sum of inputs
    weighted_sum = sum(i * w for i, w in zip(inputs, weights))

    # Apply threshold activation function
    if weighted_sum >= threshold:
        return 1
    else:
        return 0

# AND GATE
print("AND Gate")

# Set weights and threshold for AND gate
weights=[1,1]
threshold=2

# Display table header
print("X1 X2 | output")

# Generate AND gate truth table
for X in inputs:
    output = mcp_neuron(X, weights, threshold)
    print(f"{X[0]}  {X[1]}  |  {output}")

print("\n")

# OR GATE
print("OR Gate")

# Set weights and threshold for OR gate
weights=[1,1]
threshold=1

# Display table header
print("X1 X2 | output")

# Generate OR gate truth table
for X in inputs:
    output = mcp_neuron(X, weights, threshold)
    print(f"{X[0]}  {X[1]}  |  {output}")
```
